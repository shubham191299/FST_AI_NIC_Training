## Step 1: Import Libraries

Gradio is a Python library used to create interactive web-based
interfaces for machine learning and AI applications.

We will use Gradio to build and launch our chatbot.

```python
# Import the Gradio library
import gradio as gr

print("Gradio version:", gr.__version__)
```

## Step 2: Define the Chatbot Function

We define a Python function that accepts the user's message
and generates an appropriate response based on predefined keywords.

The chatbot covers various concepts related to Artificial Intelligence,
Machine Learning, Deep Learning, Natural Language Processing (NLP),
and Generative AI.

Conditional statements are used to identify keywords and
generate relevant responses.

```python
def chatbot_response(message, history):

    # Convert user input to lowercase for easy comparison
    message = message.lower()

    # Artificial Intelligence
    if "artificial intelligence" in message or " ai " in f" {message} ":
        return "Artificial Intelligence (AI) is a branch of computer science that enables machines to perform tasks requiring human intelligence, such as reasoning, learning, and decision-making."

    # Machine Learning
    elif "machine learning" in message:
        return "Machine Learning (ML) is a subset of AI that enables computers to learn patterns from data and make predictions without being explicitly programmed."

    # Deep Learning
    elif "deep learning" in message:
        return "Deep Learning is a subset of Machine Learning that uses artificial neural networks with multiple layers to learn complex patterns from large datasets."

    # Neural Networks
    elif "neural network" in message:
        return "Neural Networks are computing systems inspired by the human brain. They consist of interconnected nodes called neurons that process information and identify patterns."

    # Transformers
    elif "transformer" in message:
        return "Transformers are deep learning architectures that use self-attention mechanisms to understand relationships between words or tokens. They are widely used in modern Large Language Models such as GPT."

    # Autoencoders
    elif "autoencoder" in message or "auto encoder" in message:
        return "Autoencoders are neural networks used for unsupervised representation learning. They encode input data into a compressed representation and then reconstruct the original input."

    # Natural Language Processing
    elif "natural language processing" in message or "nlp" in message:
        return "Natural Language Processing (NLP) is a field of AI that enables computers to understand, interpret, and generate human language."

    # Generative AI
    elif "generative ai" in message:
        return "Generative AI refers to AI systems that can create new content, including text, images, audio, video, and code."

    # Large Language Models
    elif "large language model" in message or "llm" in message:
        return "Large Language Models (LLMs) are AI models trained on large amounts of text data to understand and generate human-like language."

    # Computer Vision
    elif "computer vision" in message:
        return "Computer Vision is a branch of AI that enables machines to interpret and understand visual information such as images and videos."

    # Reinforcement Learning
    elif "reinforcement learning" in message:
        return "Reinforcement Learning is a Machine Learning technique in which an agent learns through interaction with an environment using rewards and penalties."

    # Python
    elif "python" in message:
        return "Python is a popular high-level programming language widely used in AI, Machine Learning, data science, and web development."

    # Chatbot introduction
    elif "hello" in message or "hi" in message:
        return "Hello! Welcome to my AI chatbot. You can ask me about AI, Machine Learning, Deep Learning, Transformers, and other related concepts."

    # Chatbot identity
    elif "your name" in message:
        return "I am an interactive AI chatbot developed using Python and Gradio."

    # Thank you
    elif "thank" in message:
        return "You're welcome! Feel free to ask me more questions about Artificial Intelligence."

    # Goodbye
    elif "bye" in message:
        return "Goodbye! Thank you for interacting with my chatbot."

    # Default response
    else:
        return "Sorry, I don't have information about that topic yet. Please ask me about AI, Machine Learning, Deep Learning, Transformers, Autoencoders, NLP, or Generative AI."
```

## Step 3: Create the Interactive Chatbot Interface

We use the Gradio `ChatInterface()` function to build
a web-based conversational chatbot.

The interface includes:

1. A descriptive title: My AI Chatbot.
2. A welcome message explaining the chatbot's functionality.
3. Sample prompts covering different AI and Machine Learning concepts.
4. A text input box where users can enter their questions.
5. An interactive chat window to display user messages
   and chatbot responses.

Users can select the sample prompts or enter their own
questions related to Artificial Intelligence and its applications.

```python
# Create an interactive chatbot interface using Gradio

demo = gr.ChatInterface(

    # Connect the chatbot response function
    fn=chatbot_response,

    # Set the chatbot title
    title="My AI Chatbot",

    # Add a welcome message and usage instructions
    description="""
    Welcome to My AI Chatbot!

    This chatbot provides basic information about various
    Artificial Intelligence and Machine Learning concepts.

    You can explore topics such as:
    - Artificial Intelligence (AI) - Machine Learning (ML) - Deep Learning - Neural Networks - Transformers - Autoencoders 
    - Natural Language Processing (NLP) - Generative AI and Large Language Models (LLMs) - Computer Vision - Reinforcement Learning

    Select a sample prompt below or type your own question
    to interact with the chatbot.
    """,

    # Provide sample prompts for users
    examples=[
        "What is Artificial Intelligence?",
        "Explain Machine Learning",
        "What is Deep Learning?",
        "How do Transformers work?",
        "What are Autoencoders?",
        "Explain Generative AI",
        "What is Natural Language Processing?",
        "What is Computer Vision?",
        "Explain Reinforcement Learning"
    ]
)
```

## Step 4: Launch the Gradio Application

The `launch()` method starts the Gradio web application.

After executing the cell, Gradio generates a local URL
through which users can access and interact with the chatbot.

The interface can be tested directly in the Jupyter Notebook
or opened in a web browser.

```python
# Launch the chatbot application
demo.launch()
```
