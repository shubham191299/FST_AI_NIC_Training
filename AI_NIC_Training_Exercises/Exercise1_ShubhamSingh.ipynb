### Step 1: Import Required Libraries

Import Pandas, NumPy and Matplotlib for data generation, 
data manipulation and visualization.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

### Step 2: Create a DataFrame

Generate random numerical values for four columns (f1, f2, f3 and f4).
We will create 10 rows of data using NumPy's random function.

```python
# Set seed to ensure reproducible random values
np.random.seed(42)

# Generate random values for four columns
data = np.random.randint(1, 100, size=(10, 4))

# Create a Pandas DataFrame
df = pd.DataFrame(data, columns=['f1', 'f2', 'f3', 'f4'])

# Display the DataFrame
df
```

### Step 3: Data Visualization – Scatter Plot

Create a scatter plot to visualize the relationship between 
columns f2 and f3.

The X-axis represents f2, while the Y-axis represents f3.

```python
# Create a scatter plot
plt.figure(figsize=(8, 5))

plt.scatter(df['f2'], df['f3'], color='blue', marker='o')

# Add chart title and axis labels
plt.title('Scatter Plot: f2 vs f3')
plt.xlabel('f2')
plt.ylabel('f3')

# Display grid and plot
plt.grid(True)
plt.show()
```

### Step 4: Calculate Row-wise Arithmetic Mean

Calculate the arithmetic mean of columns f1, f2, f3 and f4 
for each row and store the result in a new column named f5.

The mean is calculated by adding the four values in each row 
and dividing their sum by 4.

```python
# Calculate row-wise mean and create a new column f5
df['f5'] = df[['f1', 'f2', 'f3', 'f4']].mean(axis=1)

# Display the updated DataFrame
df
```

### Step 5: Export DataFrame to Excel

Export the updated DataFrame containing all five columns 
(f1, f2, f3, f4 and f5) into a Microsoft Excel file.

```python
# Export DataFrame to Excel
df.to_excel('Exercise1_Output.xlsx', index=False)

print("DataFrame successfully exported to Excel!")
```
