
# Exercise 3: Generative AI — Prompt Engineering for Video Synthesis

## Objective
To formulate structured and effective prompts for multimodal generative AI tools and generate a short video using multiple image inputs.

## Introduction
Generative AI tools can create videos from text prompts and reference images. In this exercise, Runway is selected as the AI video-generation tool to create a 3-second video incorporating two original images and their blended output from Exercise 2.

The video will demonstrate a smooth visual transition between the two source images and the blended image.


## 1. Selected AI Video-Generation Tool

**Tool Name:** Runway AI

**Website:** https://runwayml.com/

**Purpose:** To generate a short video using multiple reference images and a structured text prompt.

**Input Images:**
1. First source image.
2. Second source image.
3. Blended image generated in Exercise 2.

**Expected Output:** A 3-second video showing a smooth transition between the three images.


## 2. Structured Prompt for Video Generation

**Exact Prompt:**

Create a cinematic 3-second video using the three provided reference images.

Start with the first source image displayed clearly for a brief moment. Gradually transition into the second source image using a smooth cross-dissolve effect. Then seamlessly blend both images into the final blended image, highlighting the combination of visual elements.

Maintain the original colors, objects, proportions, and visual details of all three reference images. Use smooth transitions without introducing additional objects or changing the original image content.

**Camera Instructions:**
- Use a fixed camera position.
- Apply a subtle zoom-in effect throughout the video.
- Maintain a stable composition.
- Avoid camera shake or sudden movements.

**Transition Instructions:**
- 0–1 second: Display the first source image.
- 1–2 seconds: Smoothly transition to the second source image.
- 2–3 seconds: Transition into the final blended image.

**Video Duration:** 3 seconds.

**Visual Style:** Realistic, smooth, cinematic, and high-quality.

**Output:** A 3-second video in which all three images are presented through seamless transitions.


## 3. Camera and Transition Instructions

| Parameter | Description |
|---|---|
| Camera Position | Fixed camera |
| Camera Movement | Subtle zoom-in |
| First Transition | First image to second image |
| Second Transition | Second image to blended image |
| Transition Effect | Smooth cross-dissolve |
| Video Duration | 3 seconds |
| Visual Quality | High quality |
| Composition | Preserve original image details |


## 4. Expected Output and Observation

**Expected Output:**

A 3-second video containing all three reference images, with smooth transitions from the first source image to the second source image and finally to the blended image.

**Observation:**

The exercise demonstrates how prompt engineering helps control the visual sequence, camera movement, transition effects, and duration of AI-generated videos.

A well-structured prompt provides clearer instructions to the generative AI model and helps produce more consistent results.
